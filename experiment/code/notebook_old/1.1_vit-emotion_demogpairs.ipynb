# Extraksi Fitur Ekspresi

```python
import utils as u
import torch
from transformers import AutoModelForImageClassification
from tqdm import tqdm
import pandas as pd
import joblib
import numpy as np

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
device
```

## Load Model

```python
model_path = 'dima806/facial_emotions_image_detection'
model = AutoModelForImageClassification.from_pretrained(model_path).to(device)
model_path
```

## DemogPairs

### Load Dataset

```python
data = u.load_demogpairs()
pd.DataFrame(data)
```

### Ektraksi Fitur

```python
features = {}
for d in tqdm(data):
    features[d['image_path']] = u.extract_vit_features(d['full_path'], model=model, model_path=model_path, device=device)
features[d['image_path']].shape, joblib.dump(features, 'features/demogpairs_vit-emotion.pkl', compress=9)
```
